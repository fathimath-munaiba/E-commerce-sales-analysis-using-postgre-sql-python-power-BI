###EDA


```python
sales=order_items_clean.merge(
    orders_clean,
    on ="order_id",
    how= "inner")
sales=sales.merge(
    products_clean,
    on="product_id",
    how ="inner")
print("sales dataset shape:",sales.shape)
```

```python
sales.info()
```

```python
sales.head()
```

```python
sales.columns.tolist()
```

```python
sales = sales.rename(columns={
    "unit_price_aed_x": "selling_price_aed",
    "unit_price_aed_y": "product_list_price_aed"
})

sales["order_date"]=pd.to_datetime(
    sales["order_date"],errors="coerce")
```

```python
sales.info()
```

```python
#revenue, profit,profit margin
```

```python
sales["revenue"]=sales["quantity"]*sales["selling_price_aed"]-sales["discount_aed"]
sales["total_cost"]=sales["quantity"]*sales["unit_cost_aed"]
sales["profit"]=sales["revenue"]-sales["total_cost"]
sales["profit_margin"]=100*(sales["profit"]/sales["revenue"])
print("sales revenue is:",sales["revenue"])
print("total_cost :",sales["total_cost"])
print("gross_profit :",sales["profit"])
print("profit margin:",sales["profit_margin"])
```

```python
sales[[
    "product_name",
    "quantity",
    "revenue",
    "total_cost",
    "profit",
    "profit_margin"
]].head()
```

```python
#DELIVERED ORDERS
```

```python
delivered_sale=sales[sales["order_status"]=="Delivered"].copy()
print("total_delivered_order:",len(delivered_sale))
```

```python
total_revenue=delivered_sale["revenue"].sum()
total_cost=delivered_sale["total_cost"].sum()
gross_profit=delivered_sale["profit"].sum()
total_order=delivered_sale["order_id"].nunique()
total_customer=delivered_sale["customer_id"].nunique()
aov=total_revenue/total_order
profit_margin=100*(gross_profit/total_revenue)
print("total_revenue:",round(total_revenue,2))
print("total_cost:",round(total_cost,2))
print("gross_profit:",round(gross_profit,2))
print("total_order:",total_order)
print("total_customer:",total_customer)
print("average_order_value:",round(aov,2))
print("profit_margin:",round(profit_margin,2),"%")
```

```python
#monthly sales analysis
```

```python
monthly_sales = (
    delivered_sale
    .groupby(delivered_sale["order_date"].dt.to_period("M"))
    .agg(
        total_revenue=("revenue", "sum"),
        total_orders=("order_id", "nunique"),
        total_quantity=("quantity", "sum")
    )
    .reset_index()
)
monthly_sales["order_date"] = (
    monthly_sales["order_date"].dt.to_timestamp()
)
monthly_sales.head(12)
```

```python
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_sales["order_date"],
    monthly_sales["total_revenue"],
    marker="o"
)
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (AED)")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
monthly_sales=monthly_sales.sort_values("order_date")
```

```python
monthly_sales["previous_revenue"] = (
    monthly_sales["total_revenue"].shift(1)
)
```

```python
## GROWTH PERCENTAGE
```

```python
monthly_sales["mom_growth"]=100*((monthly_sales["total_revenue"]-monthly_sales["previous_revenue"])/monthly_sales["previous_revenue"])
```

```python
#CUSTOMER ANALYSIS
```

```python
customer_analysis=delivered_sale.groupby("customer_id").agg(
    total_revenue=("revenue","sum"),
    total_orders=("order_id","nunique"),
    total_quantity=("quantity","sum")).reset_index()
top_10_customers=customer_analysis.sort_values(
    by="total_revenue",
    ascending=False)

top_10_customers.head(10).round(2)    
```

```python
#ONE TIME VS REPEAT CUSTOMERS
```

```python
customer_analysis["customer_type"]=np.where(
    customer_analysis["total_orders"]==1 ,
    "One-time",
    "Repeat")
```

```python
customer_summary=customer_analysis.groupby("customer_type").agg(
    total_revenue=("total_revenue","sum"),
    total_customers=("customer_id","count"),
    average_revenue=("total_revenue","mean")
).reset_index()
customer_summary.head().round(2)
```

```python
customer_summary["repeat%"]=100*(customer_summary["total_customers"]/customer_summary["total_customers"].sum())
customer_summary.head().round(2)
```

```python
#CUSTOMER VALUE ANALYSIS
```

```python
customer_analysis["value_type"]=np.where(
    customer_analysis["total_revenue"] > 10000,
    "High_value",
    "Regular")
```

```python
customer_analysis.head().round(2)
```

```python
customer_value_summary=customer_analysis.groupby("value_type").agg(
    total_revenue=("total_revenue","sum"),
    total_customers=("customer_id","count"),
    average_revenue=("total_revenue","mean")
).reset_index()
customer_value_summary.head().round(2)
```
